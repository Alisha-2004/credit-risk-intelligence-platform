```python
from groq import Groq
from dotenv import load_dotenv
import os

load_dotenv()

client = Groq(
    api_key=os.getenv("GROQ_API_KEY")
)

response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {
            "role":"user",
            "content":"What is credit risk?"
        }
    ]
)

print(
    response.choices[0].message.content
)
```

```python
import sqlite3
import pandas as pd

df = pd.read_csv(r"C:\Users\pullo\Downloads\home-credit-default-risk\application_train.csv")

conn = sqlite3.connect("../data/credit_risk.db")

df.to_sql(
    "applications",
    conn,
    if_exists="replace",
    index=False
)

conn.close()

print("Database created")
```
